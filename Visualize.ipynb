### Vertical Graph

```python
# Using Percentage Transformation for visualization
# Sumup Total peptide in each sheet
total_peptides = sam1g1_df['nPepSeq'].sum()
sorted_df = sam1g1_df.sort_values(by='nPepSeq', ascending=False)

top10g1_df = sorted_df.head(10).copy()
top10g1_df['Hover_Details'] = ""
restg1_df = sorted_df.iloc[10:]

if not restg1_df.empty:
    others_sum = restg1_df['nPepSeq'].sum()
    others_names = restg1_df['Bioactivity'].tolist() #extract the names of the remaining categories

    if len(others_names) > 5:
        visible_name = '<br>'.join(others_names[:5])
        details_str = f"<br><b>Includes:</b><br>{visible_name}<br><i> ...and {len(others_names)-5} more</i>"
    else:
        visible_names = "<br>".join(others_names)
        details_str = f"<br><b>Includes:</b><br>{visible_names}"

    others_row = pd.DataFrame({
        'Bioactivity': ['Others'], 
        'nPepSeq': [others_sum],
        'Hover_Details': [details_str]
    })
    
    # Combine the Top 10 with the new Others row using pd.concat
    plot_df = pd.concat([top10g1_df, others_row], ignore_index=True)
else:
    # If the sheet happens to have 10 or fewer categories total, just use it as is
    plot_df = top10g1_df

plot_df['nPepSeq_percentage'] = (plot_df['nPepSeq'] / total_peptides) * 100
plot_df['Hover_Percentage'] = plot_df['nPepSeq_percentage'].round(2).astype(str) + '%'

formatted_total = f"{total_peptides:,}" # format total peptide เพื่อความสวยงาม

fig = px.bar(plot_df, 
             x='Bioactivity', 
             y='nPepSeq', 
             log_y=True,      # The Y-axis now scales by percentage
             text='nPepSeq',
             custom_data=['Hover_Percentage', 'Hover_Details'],  # Include the hover data for percentage and details
#             hover_data={
#                 'Hover_Percentage': True,     # Show percentage rounded to 2 decimals when hovering
 #                'Hover_Details': True  # Show details of the "Others" category when hovering
  #           },
             orientation='v',
             title='Top 10 Bioactivity Peptide Sequence Count',
             height=600, width=800)

# 5. Make it look clean by putting the text just above the bars and renaming the axis
fig.update_traces(textposition='outside',
                  hovertemplate='<b>%{x}</b><br>Count: %{y}<br>Percentage: %{customdata[0]} %{customdata[1]}<extra></extra>')
fig.update_layout(yaxis_title='Percentage of Total Dataset (%)')

fig.show()
```

# Test automate function

```python
import pandas as pd
import plotly.express as px
import os

def generate_dashboard_charts_multisheet(file_paths):
    """
    Iterates through a list of Excel files, opens EVERY sheet inside them, 
    and generates Bar and Pie charts for sheets that contain the correct data.
    """
    
    # 1 loop through every file in the folder
    for file in file_paths:
        file_name = os.path.basename(file)
        
        # Load ALL sheets from the file into a dictionary
        all_sheets = pd.read_excel(file, sheet_name=None)
        
        # 2 loop through every sheet
        for sheet_name, df in all_sheets.items():
            
            # If a sheet is just a cover page or notes, it skips it so the code won't crash.
            if 'nPepSeq' not in df.columns or 'Bioactivity' not in df.columns:
                print(f"Skipping '{sheet_name}' in {file_name}: Required columns not found.")
                continue  
            
            # 3  Data Processing
            # Sumup Total peptide in each sheet
            total_peptides = df['nPepSeq'].sum()
            sorted_df = df.groupby('Bioactivity', as_index=False)['nPepSeq'].sum().sort_values(by='nPepSeq', ascending=False) # Group by 'Bioactivity' and sum 'nPepSeq', then sort

            top10_df = sorted_df.head(10).copy()
            rest_df = sorted_df.iloc[10:]


            # 4 handle the rest of the data for the "Others" category
            if not rest_df.empty:
                others_sum = rest_df['nPepSeq'].sum()
                others_names = rest_df['Bioactivity'].tolist() #extract the names of the remaining categories

                if len(others_names) > 5:
                    visible_name = '<br>'.join(others_names[:5])
                    details_str = f"<br><b>Includes:</b><br>{visible_name}<br><i> ...and {len(others_names)-5} more</i>"
                else:
                    visible_names = "<br>".join(others_names)
                    details_str = f"<br><b>Includes:</b><br>{visible_names}"

                others_row = pd.DataFrame({
                    'Bioactivity': ['Others'], 
                    'nPepSeq': [others_sum],
                    'Hover_Details': [details_str]
                })
                # Combine the Top 10 with the new Others row using pd.concat
                plot_df = pd.concat([top10_df, others_row], ignore_index=True)
            else: 
                plot_df = top10_df    

            plot_df['Hover_Percentage'] = ((plot_df['nPepSeq'] / total_peptides) * 100).round(2).astype(str) + '%'             
            plot_df['Hover_Combined'] = plot_df['Hover_Percentage'] + plot_df['Hover_Details']
            formatted_total = f"{total_peptides:,}"


            # 5 Generate the Bar Chart
            fig_bar = px.bar(plot_df, 
                             x='nPepSeq', 
                             y='Bioactivity', 
                             log_x=True,      # The Y-axis now scales by percentage
                             text='nPepSeq',
                             custom_data=['Hover_Percentage', 'Hover_Details'],  # Include the hover data for percentage and details
                             orientation='h',
                             title=f'Top 10 Bioactivity (Bar) - {file_name} | Sheet: {sheet_name}<br><sup>Total Peptide Sequences: {formatted_total}</sup>',
                             height=600, width=1000)

            fig_bar.update_traces(textposition='outside',hovertemplate='<b>%{y}</b><br>Count: %{x}<br>Percentage: %{customdata[0]}%{customdata[1]}<extra></extra>')
            fig_bar.update_layout(
                                  yaxis=dict(autorange="reversed", title='Bioactivity'),
                                  xaxis=dict(title='Count of Peptide Sequences (Log Scale)')
                                  )
            
            fig_bar.show()

            # 6 Generate the Pie Chart
            fig_pie = px.pie(plot_df, 
                             values='nPepSeq', 
                             names='Bioactivity', 
                             hover_name='Bioactivity',
                             custom_data=['Hover_Combined'],  # Include the combined hover data
                             title=f'Top 10 Bioactivity (Bar) - {file_name} | Sheet: {sheet_name}<br><sup>Total Peptide Sequences: {formatted_total}</sup>',
                             height=600, width=800)
            
            fig_pie.update_traces(textposition='inside', 
                                  textinfo='percent+label',
                                  hovertemplate='<b>%{label}</b><br>Count: %{value}<br>%{customdata[0]}<extra></extra>'
            )
            
            fig_pie.show()
```

```python
my_files = ['sample1.xlsx', 'sample2.xlsx', 'sample3.xlsx']  # Add your Excel file paths here
generate_dashboard_charts_multisheet(my_files)
```
